```python
import numpy as np
import pandas as pd
```

```python
# Read Data

df = pd.read_excel("/content/FlavonoidData.xlsx")
df
```

```python
# Discretized Data

df['v01'] = 1
df.loc[df['viability-mean']>50, 'v01'] = 0
```

```python
df = df.drop(['authors', 'viability-error', 'viability-SD', 'viability-SE'], axis=1)
df
```

```python
# Encoded Data

df_encoded = pd.get_dummies(df, dtype=int)
df_encoded
```
