```python
import numpy as np
import pandas as pd
```

```python
# Read Data

df = pd.read_excel("/content/CV.xlsx")
df
```

```python
# Select "Does=40" Rows

df_40 = df[df['dose']==40]
df_40
```

```python
# 1st, 2nd, and 3rd Column

df_40[['authors', 'flavonoid', 'cell line', 'time', 'viability-mean']].groupby(['authors', 'flavonoid', 'cell line']).size()
```

```python
# Mean for Two Last Columns

mean_ = df_40[['flavonoid', 'cell line', 'time', 'viability-mean']].groupby(['flavonoid', 'cell line', 'time']).mean()
mean_
```

```python
# STD for Two Last Columns

std_ = df_40[['flavonoid', 'cell line', 'time', 'viability-mean']].groupby(['flavonoid', 'cell line', 'time']).std()
std_
```

```python
# Two Last Columns

std_ / mean_
```
